```python
print('''
=================================================================================================================================== 
                                              SMART BUS TICKET BOOKING SYSTEM 
=================================================================================================================================== 
                                                    How Can I help you!
                                            ===================================
                    1. View Routes                                   2. Search Route
                    3. Book Ticket                                   4. View Bookings
                    5. Cancel Booking                                6. Check Seats
                    7. Booking Report                                8. Exit
''')
routes=[{'id':1001,'from':'Karachi','to':'Peshawer','price':3500,'seats':50},
        {'id':1002,'from':'Karachi','to':'Lahore','price':2500,'seats':45},
       {'id':1003,'from':'Lahore','to':'Islamabad','price':2000,'seats':35},
       {'id':1004,'from':'Karachi','to':'Balakot','price':3200,'seats':20},
       {'id':1005,'from':'Kashmir','to':'Quetta','price':5000,'seats':30} ]
bookings = []
booking_ids = set()
next_id = 1001

for menu in range(1000):
    choice=int(input('Enter your Choice: '))
    if choice==1:
        for route in routes:
            print(route['id'],':',route['from'],'to' ,route['to'],'Price:',route['price'],'Seats:',route['seats'])
    if choice==2:
        start=input('Enter departure city:')
        end=input('Enter arrival city:')
        found=False
        for route in routes:
            if start==route['from'] and end==route['to']:
                print('Route Found!')
                print(f'Price:{route['price']}: Seats:{route['seats']}')
                found=True
        if found==False:
            print('Record not found')
    if choice==3:
        start = input("From city: ").title()
        end = input("To city: ").title()
        selected = None

        for route in routes:
            if route["from"] == start and route["to"] == end:
                selected = route

        if selected == None:
            print("Route not found.")
        else:
            name = input("Passenger name: ")
            age = input("Passenger age: ")
            quantity = input("Number of tickets: ")

            if not age.isdigit() or not quantity.isdigit():
                print("Please enter numbers for age and tickets.")
            elif int(age) <= 0 or int(quantity) <= 0:
                print("Age and tickets must be greater than zero.")
            elif int(quantity) > selected["seats"]:
                print("Not enough seats available.")
            else:
                booking_id = next_id
                next_id += 1
                booking_ids.add(booking_id)

                total = selected["price"] * int(quantity)
                selected["seats"] -= int(quantity)

                booking = {
                    "id": booking_id,
                    "name": name,
                    "age": int(age),
                    "from": start,
                    "to": end,
                    "tickets": int(quantity),
                    "total": total,
                    "cancelled": False
                }

                bookings.append(booking)

                print("Booking successful!")
                print("Booking ID:", booking_id)
                print("Total fare:", total)
    if choice==4:
        if len(bookings) == 0:
            print("No bookings found.")
        else:
            for booking in bookings:
                print("--------------------")
                print("Booking ID:", booking["id"])
                print("Name:", booking["name"])
                print("Age:", booking["age"])
                print("Route:", booking["from"], "to", booking["to"])
                print("Tickets:", booking["tickets"])
                print("Total fare:", booking["total"])

                if booking["cancelled"]:
                    print("Status: Cancelled")
                else:
                    print("Status: Confirmed")
    if choice==5:
        cancel_id = input("Enter booking ID: ")

        if not cancel_id.isdigit():
            print("Invalid booking ID.")
        else:
            found = False
            for booking in bookings:
                if booking["id"] == int(cancel_id):
                    found = True
                    if booking["cancelled"]:
                        print("Booking already cancelled.")
                    else:
                        booking["cancelled"] = True
                        for route in routes:
                            if route["from"] == booking["from"] and route["to"] == booking["to"]:
                                route["seats"] += booking["tickets"]
                        print("Booking cancelled successfully.")
            if found == False:
                print("Booking not found.")
    if choice==6:
        for route in routes:
            print(route["from"], "to", route["to"],
                  "Available seats:", route["seats"])
    if choice==7:
        total_bookings = 0
        total_tickets = 0
        revenue = 0

        for booking in bookings:
            if booking["cancelled"] == False:
                total_bookings += 1
                total_tickets += booking["tickets"]
                revenue += booking["total"]

        print("Confirmed bookings:", total_bookings)
        print("Tickets sold:", total_tickets)
        print("Total revenue:", revenue)
    if choice==8:
        print('Thank you for using Bus booking Application')
        break
```
